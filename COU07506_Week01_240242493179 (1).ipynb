```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
from scipy.io import wavfile

plt.rcParams["figure.figsize"] = (9, 3.5)

print("DSP Laboratory is ready!")
```

```python
n = np.arange(-5, 11)

delta = (n == 0).astype(float)
step = (n >= 0).astype(float)
ramp = n * step
expo = (0.8**n) * step

signals = [delta, step, ramp, expo]
titles = ["Unit Impulse", "Unit Step",
          "Unit Ramp", "Exponential"]

fig, axes = plt.subplots(1, 4, figsize=(15, 4))

for i in range(4):
    axes[i].stem(n, signals[i])
    axes[i].set_title(titles[i])
    axes[i].set_xlabel("n")
    axes[i].set_ylabel("Amplitude")
    axes[i].grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(0, 21)

a_values = [0.5, -0.8, 1.1]

fig, axes = plt.subplots(1, 3, figsize=(14, 4))

for i, a in enumerate(a_values):
    x = a ** n
    axes[i].stem(n, x)
    axes[i].set_title(f"a = {a}")
    axes[i].set_xlabel("n")
    axes[i].set_ylabel("Amplitude")
    axes[i].grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(0, 41)

x1 = np.cos(2 * np.pi * n / 8)
x2 = np.cos(2 * np.pi * n / 7.5)

fig, axes = plt.subplots(2, 1, figsize=(12, 7))

axes[0].stem(n, x1)
axes[0].set_title("Cosine Signal: Period = 8")
axes[0].set_xlabel("n")
axes[0].set_ylabel("Amplitude")
axes[0].grid(True)

axes[1].stem(n, x2)
axes[1].set_title("Cosine Signal: Period = 7.5")
axes[1].set_xlabel("n")
axes[1].set_ylabel("Amplitude")
axes[1].grid(True)

plt.tight_layout()
plt.show()
```

```python
from scipy.io import wavfile

fs, audio = wavfile.read("/backend-container/myvoice.wav")

print("Sampling Frequency:", fs)
print("Duration:", len(audio) / fs, "seconds")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Convert audio samples into time
time = np.arange(len(audio)) / fs

# Plot audio waveform
plt.figure(figsize=(12, 4))
plt.plot(time, audio)

plt.title("My Recorded Audio Signal")
plt.xlabel("Time (Seconds)")
plt.ylabel("Amplitude")
plt.grid(True)

plt.show()
```

```python
# Convert stereo audio to mono if necessary
if audio.ndim == 2:
    audio = audio.mean(axis=1)

# Convert audio to float
audio = audio.astype(float)

# Normalize audio
if np.max(np.abs(audio)) != 0:
    audio = audio / np.max(np.abs(audio))

# Plot normalized audio
time = np.arange(len(audio)) / fs

plt.figure(figsize=(12, 4))
plt.plot(time, audio)

plt.title("Normalized Audio Signal")
plt.xlabel("Time (Seconds)")
plt.ylabel("Normalized Amplitude")
plt.grid(True)

plt.show()
```
